```python
import pandas as pd
import requests
```

```python
url = "https://www.datos.gov.co/resource/p6dx-8zbt.json"
params = {"$limit": 5000}

response = requests.get(url, params=params)
print(response.status_code) 
```

```python
datos = response.json()
df = pd.DataFrame(datos) 
```
